```python
import pandas as pd
```

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'

column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']

iris_df = pd.read_csv(url, names=column_names)
```

```python
iris_df.head()
```

```python
iris_df.describe()
```

```python
mean_speal_lenth = iris_df['sepal_length'].mean()
print(mean_speal_lenth)
```

```python
max_speal_lenth = iris_df['sepal_length'].max()
print(max_speal_lenth)
```

```python
mean_per_speciies = iris_df.groupby('species').mean()
mean_per_speciies
```
